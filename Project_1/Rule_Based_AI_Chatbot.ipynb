```python

# Project 1: Rule-Based AI Chatbot
# Created by: Nwonu Ikenna Basil

print("===================================")
print("       WELCOME TO MY CHATBOT       ")
print("===================================")

print("Chatbot: Hello! I am your AI chatbot.")
print("Chatbot: You can chat with me.")
print("Chatbot: Type 'bye', 'exit', or 'quit' to end the conversation.")

while True:

    user_input = input("\nYou: ").lower().strip()

    # Greeting responses
    if user_input in ["hello", "hi", "hey"]:
        print("Chatbot: Hello! Nice to meet you.")

    elif user_input == "good morning":
        print("Chatbot: Good morning! Have a wonderful day.")

    elif user_input == "good afternoon":
        print("Chatbot: Good afternoon! How is your day going?")

    elif user_input == "good evening":
        print("Chatbot: Good evening! I hope you had a great day.")

    # General conversation
    elif user_input == "how are you":
        print("Chatbot: I am doing great! Thanks for asking.")

    elif user_input == "what is your name":
        print("Chatbot: I am a Rule-Based AI Chatbot.")

    elif user_input == "who created you":
        print("Chatbot: I was created by Nwonu Ikenna Basil.")

    elif user_input == "what can you do":
        print("Chatbot: I can respond to simple predefined messages.")

    elif user_input == "thank you":
        print("Chatbot: You're welcome! Happy to help.")

    # Basic AI knowledge
    elif user_input == "what is ai":
        print("Chatbot: AI stands for Artificial Intelligence. It enables machines to perform tasks that normally require human intelligence.")

    elif user_input == "what is python":
        print("Chatbot: Python is a popular programming language used in AI, machine learning, and software development.")

    # Exit commands
    elif user_input in ["bye", "exit", "quit"]:
        print("Chatbot: Goodbye! It was nice talking to you.")
        break

    # Unknown inputs
    else:
        print("Chatbot: Sorry, I don't understand that. Can you try something else?")
```
