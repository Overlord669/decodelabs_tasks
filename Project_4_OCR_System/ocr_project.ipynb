```python

# STEP 1: Prepare the image
ocr_image = image.convert("L")

# STEP 2: Apply image preprocessing
processed_image = ocr_image.filter(
    ImageFilter.GaussianBlur(radius=1)
)

# STEP 3: Extract OCR information
ocr_results = pytesseract.image_to_data(
    processed_image,
    output_type=pytesseract.Output.DICT
)

# STEP 4: Set confidence threshold
confidence_threshold = 80

# STEP 5: Create a copy for visualisation
final_image = image.copy()
draw = ImageDraw.Draw(final_image)

valid_confidences = []
accepted_words = []

# STEP 6: Analyse recognised words
for i in range(len(ocr_results["text"])):

    word = ocr_results["text"][i].strip()
    confidence = float(ocr_results["conf"][i])

    if word and confidence >= 0:

        valid_confidences.append(confidence)

        if confidence >= confidence_threshold:

            accepted_words.append(word)

            x = ocr_results["left"][i]
            y = ocr_results["top"][i]
            w = ocr_results["width"][i]
            h = ocr_results["height"][i]

            # Draw bounding box
            draw.rectangle(
                [x, y, x + w, y + h],
                outline="red",
                width=2
            )

# STEP 7: Calculate average confidence
if valid_confidences:
    average_confidence = np.mean(valid_confidences)
else:
    average_confidence = 0

# STEP 8: Display results
print("========== OCR PROJECT RESULTS ==========")

print("Recognised text:")
print(pytesseract.image_to_string(processed_image))

print("\nAccepted words (80% and above):")
print(accepted_words)

print("\nAverage confidence:",
      round(average_confidence, 2), "%")

print("\nConfidence requirement:")

if average_confidence >= confidence_threshold:
    print("PASSED: Average confidence meets the 80% target.")
else:
    print("NOT MET: Average confidence is below 80%.")

print("\nNumber of accepted words:", len(accepted_words))

# STEP 9: Display final visual result
plt.figure(figsize=(12, 6))
plt.imshow(final_image)
plt.title("Final OCR Recognition Results")
plt.axis("off")
plt.show()

```

```python
import pytesseract
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Tesseract version:", pytesseract.get_tesseract_version())
print("All required libraries imported successfully!")
```

```python

# ==========================================
# DECODELABS PROJECT 4
# COMPLETE OCR PROCESSING PIPELINE
# ==========================================

# STEP 1: Convert image to grayscale
gray = cv2.cvtColor(
    np.array(real_image),
    cv2.COLOR_RGB2GRAY
)

# STEP 2: Resize image
resized = cv2.resize(
    gray,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_CUBIC
)

# STEP 3: Apply adaptive thresholding
binary = cv2.adaptiveThreshold(
    resized,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    31,
    10
)

# STEP 4: Perform OCR
ocr_results = pytesseract.image_to_data(
    binary,
    config="--psm 6",
    output_type=pytesseract.Output.DICT
)

# STEP 5: Extract words and confidence scores
detected_words = []
accepted_words = []
confidence_scores = []

confidence_threshold = 80

for i, word in enumerate(ocr_results["text"]):

    word = word.strip()
    confidence = float(ocr_results["conf"][i])

    if word and confidence >= 0:
        detected_words.append(word)
        confidence_scores.append(confidence)

        if confidence >= confidence_threshold:
            accepted_words.append(word)

# STEP 6: Calculate average confidence
if confidence_scores:
    average_confidence = np.mean(confidence_scores)
else:
    average_confidence = 0

# STEP 7: Draw bounding boxes
final_image = cv2.cvtColor(
    binary,
    cv2.COLOR_GRAY2RGB
)

for i, word in enumerate(ocr_results["text"]):

    word = word.strip()
    confidence = float(ocr_results["conf"][i])

    if word and confidence >= confidence_threshold:

        x = ocr_results["left"][i]
        y = ocr_results["top"][i]
        w = ocr_results["width"][i]
        h = ocr_results["height"][i]

        cv2.rectangle(
            final_image,
            (x, y),
            (x + w, y + h),
            (255, 0, 0),
            2
        )

# STEP 8: Display final report
print("========== OCR PIPELINE REPORT ==========")

print("\nExtracted Text:")
print(" ".join(detected_words))

print("\nAccepted Words:")
print(" ".join(accepted_words))

print("\nTotal Detected Words:", len(detected_words))
print("Words Above 80% Confidence:", len(accepted_words))

print("Average Confidence:",
      round(average_confidence, 2), "%")

if average_confidence >= confidence_threshold:
    print("Overall Result: PASSED")
else:
    print("Overall Result: BELOW THRESHOLD")

# STEP 9: Display final detection image
plt.figure(figsize=(14, 8))
plt.imshow(final_image)
plt.title("Complete OCR Pipeline - Detected Text")
plt.axis("off")
plt.show()
```

```python

import pandas as pd

# Tesseract configurations
psm_modes = {
    "PSM 3": "--oem 3 --psm 3",
    "PSM 6": "--oem 3 --psm 6",
    "PSM 11": "--oem 3 --psm 11"
}

results = []

for mode_name, config in psm_modes.items():

    prediction = pytesseract.image_to_string(
        binary,
        config=config
    )

    word_error = wer(ground_truth, prediction)
    character_error = cer(ground_truth, prediction)

    results.append({
        "Configuration": mode_name,
        "Word Error Rate (%)": round(word_error * 100, 2),
        "Character Error Rate (%)": round(character_error * 100, 2),
        "Word Accuracy (%)": round((1 - word_error) * 100, 2),
        "Character Accuracy (%)": round((1 - character_error) * 100, 2)
    })

# Create comparison table
results_df = pd.DataFrame(results)

display(results_df)

# Identify best configuration by word error rate
best_result = results_df.loc[
    results_df["Word Error Rate (%)"].idxmin()
]

print("\nBEST PERFORMING CONFIGURATION")
print(best_result)
```

```python

plt.figure(figsize=(10, 5))

plt.bar(
    results_df["Configuration"],
    results_df["Word Accuracy (%)"],
    label="Word Accuracy"
)

plt.plot(
    results_df["Configuration"],
    results_df["Character Accuracy (%)"],
    marker="o",
    label="Character Accuracy"
)

plt.title("OCR Performance Comparison")
plt.xlabel("Tesseract Configuration")
plt.ylabel("Accuracy (%)")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```
